## Introducción 

Al trabajar conseries de tiempo, a menudo enfrentamos el desafío de comprender y pronosticar patrones en datos que evolucionan con el tiempo. 


- las ventas mensuales de una empresa,
-  la temperatura diaria de una ciudad
- el número de pacientes que llegan a un hospital cada semana. 


El suavizamiento exponencial es una técnica poderosa y sencilla que  ayuda a lograr esa tarea. A diferencia de otros métodos, como las medias móviles, el suavizamiento exponencial da mayor peso a las observaciones más recientes, lo que permite adaptarse mejor a cambios o patrones recientes en los datos. Es como un "recuerdo selectivo" que enfatiza lo nuevo sin olvidar por completo el pasado.

## ¿Por qué es útil?

### Simplicidad y eficacia:
- No requiere suposiciones complejas ni modelos matemáticos avanzados. Es un punto de **entrada**  al mundo del pronóstico.
- Adaptabilidad a diferentes patrones:
Podemos ajustar el modelo para captar niveles, tendencias y estacionalidades, lo que lo convierte en una herramienta versátil para distintos contextos.

|

- Técnica para suavizar datos de series de tiempo utilizando una función de ventana exponencial. 

- Difiere  del promedio móvil simple, con el tiempo las funciones exponenciales asignan pesos exponencialmente decrecientes. 
  + Los pesos mayores se asignan a los valores u observaciones recientes, mientras que los pesos menores se asignan a los valores u observaciones más antiguos. 
  
- Entre muchas funciones de ventana, en el procesamiento de señales, la función de suavizado exponencial generalmente se aplica para suavizar datos donde actúa como un filtro de paso bajo para eliminar el ruido de alta frecuencia.

La suavización exponencial de Holt-Winters es una técnica de pronóstico para series temporales con tendencia y estacionalidad. Combina cuatro enfoques fundamentales:

![](https://miro.medium.com/v2/resize:fit:720/format:webp/1*X_Y6h0-00wDw_vV1V3CJwQ.png)

**Promedios Ponderados:**
Se calculan a partir de un conjunto de valores históricos, asignando diferentes pesos según una función específica (logarítmica, lineal, cuadrática, cúbica o exponencial). El objetivo es dar mayor relevancia a ciertos valores, generalmente los más recientes.

**Suavizado Exponencial (ES):**
Estima el próximo valor mediante un promedio ponderado de todos los datos previos, con pesos que disminuyen exponencialmente en el tiempo. Es adecuado para datos sin tendencia ni estacionalidad.

**Suavizado Exponencial de Holt:**
Extiende el método ES al incorporar una componente de tendencia, útil para series con crecimiento o decrecimiento.

**Suavizado Exponencial de Holt-Winters:**
Añade al método de Holt una componente estacional, permitiendo pronósticos en series con patrones recurrentes.

### **Promedios móviles (simples de orden k)**
$$F_{t+1} =\frac{Y_t + Y_{t-1} +……+ Y_{t-k+1}}{k}$$

- $Y_t$: observación en el período t 
- $F_t$ : pronóstico para el período t


El método de los promedios móviles utiliza el promedio de los $k$
valores de datos más recientes en la serie de tiempo como el
pronóstico para el siguiente periodo.

El término móvil indica que, mientras se dispone de una nueva
observación para la serie de tiempo, reemplaza a la observación más
antigua de la ecuación anterior y se calcula un promedio nuevo.

Como resultado, el promedio cambiará, o se moverá, conforme surjan
nuevas observaciones.


**Ejemplo:** Promedios móviles (simples de orden 3)
$$F_{t+1} =\frac{Y_t + Y_{t-1} + Y_{t-2}}{3}$$

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()

# Datos
semanas = np.arange(1, 13)
valores = np.array([17, 21, 19, 23, 18, 16, 18, 20, 22, 20, 15, 22])

k = 3  # tamaño de ventana

# Crear DataFrame
df = pd.DataFrame({
    "Semana": semanas,
    "Valor": valores
})

# Calcular promedio móvil
df["Pronostico"] = df["Valor"].rolling(window=k).mean().shift(1)

# Calcular error
df["Error"] = df["Valor"] - df["Pronostico"]

# Error cuadrático
df["Error_cuadrado"] = df["Error"]**2

df
````

````python
MAE = df["Error"].abs().mean()
MSE = df["Error_cuadrado"].mean()
RMSE = np.sqrt(MSE)

print("MAE:", MAE)
print("RMSE:", RMSE)
````

### **Promedios Móviles   Ponderados**

En el método de promedios móviles, cada observación en el cálculo recibe el mismo peso.

- Una variación, conocida como promedios móviles ponderados, consiste en seleccionar diferentes pesos para cada valor de datos y luego calcular un
promedio ponderado de los k valores de datos más recientes como el pronóstico.

- En la mayoría de los casos la observación más reciente recibe el mayor peso, y el peso disminuye para los valores de datos más antiguos. Por ejemplo, para la serie de tiempo de las venta de nafta semanal el cálculo de un promedio móvil ponderado de tres
semanas, donde la observación más reciente recibe un peso del triple del peso dado a la observación más antigua y la siguiente
observación más antigua recibe un peso del doble que la observación más antigua.

- Para la semana 4 el cálculo es:

$$\frac{3}{6}\times 19+\frac{2}{6} \times 21+\frac{1}{6} \times 17=19.33$$

En general, si creemos que el pasado reciente es un mejor
pronosticador del futuro que el pasado distante, los pesos más
grandes deben darse a las observaciones más recientes.

````python
#@title  Exposiciones por H3 — parametrizable (binarias, numéricas, ordinales)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# 1) Datos
# -----------------------------
semanas = np.arange(1, 13)
valores = np.array([17, 21, 19, 23, 18, 16, 18, 20, 22, 20, 15, 22])

# -----------------------------
# 2) Configuración WMA
#    Ejemplo de 3 semanas con pesos: 1 (más antiguo), 2, 3 (más reciente)
# -----------------------------
k = 3
pesos = np.array([1, 2, 3], dtype=float)

# Validaciones
if len(pesos) != k:
    raise ValueError("La longitud de 'pesos' debe ser igual a k.")
if np.any(pesos < 0):
    raise ValueError("Los pesos deben ser no negativos.")
if pesos.sum() == 0:
    raise ValueError("La suma de pesos debe ser > 0.")

pesos = pesos / pesos.sum()  # normalizar para que sumen 1

# -----------------------------
# 3) Función WMA (promedio ponderado de los k valores previos)
#    Nota: se usa solo pasado -> pronóstico 1-step ahead
# -----------------------------
def wma_forecast(series, k, weights):
    series = np.asarray(series, dtype=float)
    n = len(series)
    f = np.full(n, np.nan)  # pronósticos

    for t in range(k, n):
        ventana_pasada = series[t-k:t]          # y_{t-k} ... y_{t-1}
        f[t] = np.dot(weights, ventana_pasada)  # suma w_i * y_{t-k+i}
    return f

pronostico = wma_forecast(valores, k, pesos)

# -----------------------------
# 4) Tabla con errores
# -----------------------------
df = pd.DataFrame({
    "Semana": semanas,
    "Valor": valores,
    "Pronostico_WMA": pronostico
})

df["Error"] = df["Valor"] - df["Pronostico_WMA"]
df["Error_cuadrado"] = df["Error"]**2
df["Error_abs"] = df["Error"].abs()

# -----------------------------
# 5) Métricas (ignorando NaN)
# -----------------------------
valid = df["Pronostico_WMA"].notna()

MAE  = df.loc[valid, "Error_abs"].mean()
MSE  = df.loc[valid, "Error_cuadrado"].mean()
RMSE = np.sqrt(MSE)

# MAPE y sMAPE (cuidado con ceros)
y_true = df.loc[valid, "Valor"].astype(float)
y_pred = df.loc[valid, "Pronostico_WMA"].astype(float)

MAPE = (np.abs((y_true - y_pred) / y_true.replace(0, np.nan))).mean() * 100
sMAPE = (np.abs(y_true - y_pred) / ((np.abs(y_true) + np.abs(y_pred)) / 2)).mean() * 100

print(df)
print("\nMétricas (WMA):")
print(f"MAE  = {MAE:.3f}")
print(f"RMSE = {RMSE:.3f}")
print(f"MAPE = {MAPE:.2f}%")
print(f"sMAPE= {sMAPE:.2f}%")

# -----------------------------
# 6) Gráfico serie vs pronóstico + barras de error
# -----------------------------
plt.figure(figsize=(10,6))
plt.plot(df["Semana"], df["Valor"], marker="o", label="Serie real")
plt.plot(df["Semana"], df["Pronostico_WMA"], marker="o", linestyle="--", label=f"WMA (k={k}, pesos={pesos.round(2)})")

# Barras verticales: real vs pronóstico
for i in range(len(df)):
    if not np.isnan(df.loc[i, "Pronostico_WMA"]):
        plt.vlines(df.loc[i, "Semana"],
                   df.loc[i, "Pronostico_WMA"],
                   df.loc[i, "Valor"],
                   linestyle="dotted")

plt.title("Promedio Móvil Ponderado (WMA) y errores")
plt.xlabel("Semana")
plt.ylabel("Valor")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
````

## Suavizamiento Exponencial Simple (SES)

El suavizamiento exponencial simple es un modelo apropiado para series de tiempo que no presentan tendencia ni estacionalidad sistemática. El modelo asume que la serie puede describirse mediante un componente de **nivel latente** que evoluciona dinámicamente en el tiempo.

A diferencia del promedio móvil, el suavizamiento exponencial asigna pesos exponencialmente decrecientes a las observaciones pasadas, otorgando mayor importancia a los datos más recientes.

---

### Ecuación de actualización del nivel

$$
\ell_t = \alpha Y_t + (1 - \alpha)\ell_{t-1}
$$

donde:

- $\ell_t$ es el nivel estimado en el tiempo $t$,
- $Y_t$ es la observación real en el tiempo $t$,
- $\alpha \in (0,1)$ es el parámetro de suavizamiento,
- $\ell_{t-1}$ es el nivel estimado en el periodo anterior.

---

### Pronóstico a un paso adelante

$$
\hat{Y}_{t+1} = \ell_t
$$

El pronóstico corresponde al nivel estimado más reciente.

---

### Interpretación dinámica

La ecuación puede reescribirse como:

$$
\ell_t = \ell_{t-1} + \alpha (Y_t - \ell_{t-1})
$$

Esto muestra que el nivel previo se ajusta en proporción al error observado:

$$
\varepsilon_t = Y_t - \ell_{t-1}
$$

El parámetro $\alpha$ controla la velocidad de adaptación del modelo:

- Si $\alpha$ es cercano a 1, el modelo reacciona rápidamente a cambios recientes.
- Si $\alpha$ es cercano a 0, el modelo es más inercial y suaviza fuertemente la serie.

---

### Expansión como promedio ponderado infinito

Desarrollando la ecuación recursivamente se obtiene:

$$
\ell_t = \alpha Y_t + \alpha(1-\alpha)Y_{t-1} + \alpha(1-\alpha)^2 Y_{t-2} + \dots
$$

Esto demuestra que el suavizamiento exponencial simple equivale a un promedio ponderado infinito con pesos que decrecen exponencialmente a medida que las observaciones se alejan en el tiempo.

---

### Inicialización

El modelo requiere un valor inicial para el nivel $\ell_1$. Algunas opciones comunes son:

- $\ell_1 = Y_1$,
- el promedio de las primeras observaciones,
- estimación conjunta mediante máxima verosimilitud.

La elección de la inicialización puede influir en los primeros pronósticos, aunque su efecto disminuye con el tiempo.

### Expansión recursiva del modelo

El modelo de suavizamiento exponencial simple puede expandirse recursivamente para mostrar su estructura como promedio ponderado infinito.

Supongamos una serie $Y_1, Y_2, Y_3, \dots$ y una inicialización $\ell_1 = Y_1$.

Para $t=2$:

$$
\ell_2 = \alpha Y_2 + (1-\alpha)\ell_1
$$

Para $t=3$:

$$
\ell_3 = \alpha Y_3 + (1-\alpha)\ell_2
$$

Sustituyendo $\ell_2$:

$$
\ell_3 = \alpha Y_3 + (1-\alpha)\left[\alpha Y_2 + (1-\alpha)Y_1\right]
$$

$$
\ell_3 = \alpha Y_3 + \alpha(1-\alpha)Y_2 + (1-\alpha)^2 Y_1
$$

Continuando el desarrollo recursivo, se obtiene la forma general:

$$
\ell_t = \alpha Y_t + \alpha(1-\alpha)Y_{t-1} + \alpha(1-\alpha)^2 Y_{t-2} + \dots
$$

Esto demuestra que el suavizamiento exponencial simple equivale a un promedio ponderado infinito con pesos que decrecen exponencialmente a medida que las observaciones se alejan en el tiempo.

A diferencia del promedio móvil simple, el modelo no utiliza una ventana finita de datos, sino que conserva memoria infinita del pasado, aunque con impacto decreciente.

### 📌 Lectura conceptual: suavizamiento como **modelo dinámico de estados**

En este curso NO vamos a tratar el suavizamiento como una “receta de promedios”.  
Lo vamos a ver como un **modelo estructural** para series univariadas, donde existen **componentes latentes (no observables)** que evolucionan en el tiempo.

En el caso general (familia ETS), el estado puede representarse como:

$$
x_t = (\ell_t, b_t, s_t)
$$

donde:

- **Nivel** $\ell_t$: línea base o “altura” de la serie.
- **Tendencia** $b_t$: velocidad de cambio del nivel.
- **Estacionalidad** $s_t$: patrón periódico que se repite (mensual, trimestral, etc.).

En el caso particular del **Suavizamiento Exponencial Simple (SES)**, el estado se reduce únicamente al nivel:

$$
x_t = \ell_t
$$

---

### Representación en forma de espacio de estados (intuición)

Para el **suavizamiento exponencial simple (SES)** (sin tendencia ni estacionalidad) el modelo puede escribirse como:

- **Ecuación de observación**

$$
y_t = \ell_{t-1} + \varepsilon_t
$$

donde la innovación se define como:

$$
\varepsilon_t = y_t - \ell_{t-1}
$$

- **Ecuación de estado (actualización del nivel)**

$$
\ell_t = \ell_{t-1} + \alpha\,\varepsilon_t
$$

Interpretación:

El nivel anterior se ajusta en proporción al error observado.  
El parámetro $\alpha$ controla la velocidad de adaptación del modelo.

- Si $\alpha$ es grande, el modelo reacciona rápidamente a cambios recientes.
- Si $\alpha$ es pequeño, el modelo es más estable y presenta mayor inercia.

✅ Esta perspectiva será fundamental cuando trabajemos la familia completa de modelos **ETS (Error–Trend–Seasonality)** como modelos de estado.

## 📘 Ejemplo Numérico: Suavizamiento Exponencial Simple (SES)  
### Constante de suavización $\alpha = 0.2$

Consideremos la siguiente serie semanal:

$$
Y_t = \{17, 21, 19, 23, 18, 16, 20, 22, 20, 15, 22, 22\}
$$

Trabajaremos el modelo en su forma estructural:

$$
\ell_t = \alpha Y_t + (1-\alpha)\ell_{t-1}
$$

donde:

- $\ell_t$ es el **nivel estimado** en el periodo $t$,
- $Y_t$ es el valor observado en el periodo $t$,
- $\alpha$ es la constante de suavización, con $0 < \alpha < 1$.

El pronóstico a un paso adelante se obtiene como:

$$
\hat{Y}_{t+1} = \ell_t
$$

---

### 🔹 Inicialización

Para iniciar el procedimiento, tomamos como valor inicial:

$$
\ell_1 = Y_1 = 17
$$

Este será nuestro primer nivel estimado.

---

### 🔹 Cálculo paso a paso

Para el periodo 2:

$$
\ell_2 = 0.2(21) + 0.8(17)
$$

$$
\ell_2 = 4.2 + 13.6 = 17.8
$$

Para el periodo 3:

$$
\ell_3 = 0.2(19) + 0.8(17.8)
$$

$$
\ell_3 = 3.8 + 14.24 = 18.04
$$

Para el periodo 4:

$$
\ell_4 = 0.2(23) + 0.8(18.04)
$$

$$
\ell_4 = 4.6 + 14.432 = 19.03
$$

El proceso continúa de manera recursiva para los periodos siguientes.

### 🔹 Pronóstico para el periodo 13

El pronóstico a un paso adelante es:

$$
\hat{Y}_{13} = \ell_{12}
$$

$$
\hat{Y}_{13} = 19.78
$$

---

### 🎯 Interpretación

- El modelo no sigue exactamente cada oscilación de la serie.
- Ajusta progresivamente el nivel en proporción al error observado.
- Con $\alpha = 0.2$, el modelo es relativamente inercial.
- $\ell_t$ representa la mejor estimación del **nivel subyacente** de la serie en el periodo $t$.

````python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import SimpleExpSmoothing

# -----------------------------
# 1) Datos (los mismos de la tabla manual)
# -----------------------------
Y = np.array([17, 21, 19, 23, 18, 16, 20, 22, 20, 15, 22, 22], dtype=float)
semanas = np.arange(1, len(Y) + 1)

alpha = 0.2  # mismo valor usado en el ejemplo manual (antes decia 0.8)

# -----------------------------
# 2) Ajuste SES con el mismo alpha e inicializacion (l_1 = Y_1)
# -----------------------------
model = SimpleExpSmoothing(
    Y, initialization_method="known", initial_level=Y[0]
).fit(smoothing_level=alpha, optimized=False)

# level[t]        -> nivel estimado l_t (estado DESPUES de observar Y_t)
# fittedvalues[t] -> ajuste dentro de muestra, Y_hat(t|t-1) = l_(t-1)
#                    (pronostico de un paso, hecho ANTES de observar Y_t)
nivel = model.level
ajuste_dentro_muestra = model.fittedvalues

# -----------------------------
# 3) Pronostico futuro (fuera de muestra) para la semana 13
# -----------------------------
y_13 = model.forecast(1)[0]  # = l_12

print("Nivel        l_t      :", np.round(nivel, 2))
print("Ajuste  Y_hat(t|t-1)   :", np.round(ajuste_dentro_muestra, 2))
print(f"Pronostico semana 13   : {y_13:.2f}")

# -----------------------------
# 4) Grafico: serie real, nivel estimado y pronostico
# -----------------------------
plt.figure(figsize=(10, 6))

plt.plot(semanas, Y, marker="o", linewidth=2, label="Serie real $Y_t$")
plt.plot(semanas, nivel, marker="s", linestyle="--", label="Nivel estimado $\\ell_t$")

plt.scatter([len(Y) + 1], [y_13], marker="X", s=120, color="red", label="Pronostico semana 13")
plt.axvline(len(Y) + 1, linestyle=":", alpha=0.5)

plt.title(f"Suavizamiento Exponencial Simple (SES) - $\\alpha={alpha}$")
plt.xlabel("Semana")
plt.ylabel("Valor")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
````

### 🔍 Nivel, ajuste (*fitted value*) y pronóstico: tres conceptos distintos

Es un error frecuente confundir estos tres objetos. Están relacionados, pero no son lo mismo:

| Concepto | Notación | Definición | En `statsmodels` |
|---|---|---|---|
| **Nivel** | $\ell_t$ | Estado interno del modelo *después* de observar $Y_t$ | `model.level[t]` |
| **Valor ajustado (fitted value)** | $\hat{Y}_{t\mid t-1}$ | Pronóstico a un paso hecho *antes* de observar $Y_t$, con información hasta $t-1$ | `model.fittedvalues[t]` |
| **Pronóstico (forecast)** | $\hat{Y}_{t+h\mid t}$ | Proyección futura, fuera de la muestra usada para ajustar | `model.forecast(h)` |

Para el SES se cumple:

$$
\hat{Y}_{t\mid t-1} = \ell_{t-1}
$$

Es decir, **`fittedvalues[t]` no es el nivel en $t$, sino el nivel del periodo anterior**. Confundir ambos lleva a errores al graficar o al interpretar qué tan "ajustado" está el modelo en el periodo actual. Verifiquemos esto numéricamente con la corrida anterior:

de la constante de suavización tienen la ventaja de ajustar rápidamente
los pronósticos cuando ocurren errores de pronóstico y por ende
permiten que el pronóstico reaccione más rápido a las condiciones
cambiantes.

## Suavizado exponencial doble

Se utiliza muestran una tendencia pero no estacionalidad es el método de suavizado exponencial doble. El suavizado exponencial doble aplica el suavizado exponencial simple dos veces, una a los datos originales y luego a los datos resultantes del suavizado exponencial simple.

## 📘 Suavizamiento Exponencial Doble (Método de Holt)

El método de Holt extiende el suavizamiento exponencial simple para series que presentan tendencia.

El modelo introduce dos componentes dinámicos:

- $\ell_t$ : nivel estimado en el tiempo $t$.
- $b_t$ : tendencia estimada en el tiempo $t$.

---

### 1️⃣ Ecuación del nivel

$$
\ell_t = \alpha Y_t + (1 - \alpha)(\ell_{t-1} + b_{t-1})
$$

donde:

- $\alpha$ es el parámetro de suavizamiento del nivel,
- $Y_t$ es el valor observado,
- $\ell_{t-1} + b_{t-1}$ representa la proyección del nivel anterior.

---

### 2️⃣ Ecuación de la tendencia

$$
b_t = \beta(\ell_t - \ell_{t-1}) + (1 - \beta)b_{t-1}
$$

donde:

- $\beta$ es el parámetro de suavizamiento de la tendencia,
- $(\ell_t - \ell_{t-1})$ es el cambio observado en el nivel.

---

### 3️⃣ Ecuación de pronóstico

$$
\hat{Y}_{t+m} = \ell_t + m b_t
$$

donde:

- $\hat{Y}_{t+m}$ es el pronóstico a $m$ pasos adelante,
- $\ell_t$ es el nivel estimado actual,
- $b_t$ es la tendencia estimada actual,
- $m$ es el horizonte de pronóstico.

---

### 🎯 Interpretación

- El nivel captura la posición actual de la serie.
- La tendencia captura la velocidad de crecimiento o decrecimiento.
- El pronóstico extrapola linealmente el nivel según la tendencia estimada.

````python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import ExponentialSmoothing
import ipywidgets as widgets
from ipywidgets import interact

# Datos ejemplo con tendencia
Y = np.array([17, 21, 19, 23, 18, 16, 20, 22, 20, 15, 22, 22], dtype=float)
t = np.arange(1, len(Y)+1)

def holt_interactivo(alpha=0.3, beta=0.1):
    
    model = ExponentialSmoothing(
        Y,
        trend='add',
        seasonal=None
    ).fit(
        smoothing_level=alpha,
        smoothing_trend=beta,
        optimized=False
    )
    
    nivel = model.level
    tendencia = model.trend
    forecast = model.forecast(5)
    
    plt.figure(figsize=(10,6))
    
    plt.plot(t, Y, marker='o', label="Serie real")
    plt.plot(t, nivel, linestyle='--', label="Nivel estimado")
    
    # Proyección futura
    t_future = np.arange(len(Y)+1, len(Y)+6)
    plt.plot(t_future, forecast, marker='x', linestyle=':', label="Pronóstico futuro")
    
    plt.title(f"Holt (α={alpha}, β={beta})")
    plt.xlabel("Tiempo")
    plt.ylabel("Valor")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

interact(holt_interactivo,
         alpha=widgets.FloatSlider(min=0.01, max=0.99, step=0.05, value=0.3),
         beta=widgets.FloatSlider(min=0.01, max=0.99, step=0.05, value=0.1));
````

## 📘 Suavizamiento Exponencial Triple (Holt-Winters)

El método de Holt-Winters extiende el modelo de Holt incorporando un tercer componente dinámico: la estacionalidad.

El modelo permite capturar simultáneamente:

- $\ell_t$ : nivel estimado
- $b_t$ : tendencia estimada
- $s_t$ : componente estacional

Existen dos variantes del modelo:

- Versión **aditiva** (cuando la estacionalidad es aproximadamente constante).
- Versión **multiplicativa** (cuando la estacionalidad crece proporcionalmente al nivel).

Sea $L$ la longitud del ciclo estacional (por ejemplo, $L=12$ para datos mensuales).

---

### 1️⃣ Ecuación del nivel

**Versión aditiva**

$$
\ell_t = \alpha (Y_t - s_{t-L}) + (1-\alpha)(\ell_{t-1} + b_{t-1})
$$

**Versión multiplicativa**

$$
\ell_t = \alpha \left(\frac{Y_t}{s_{t-L}}\right) + (1-\alpha)(\ell_{t-1} + b_{t-1})
$$

---

### 2️⃣ Ecuación de la tendencia

$$
b_t = \beta(\ell_t - \ell_{t-1}) + (1-\beta)b_{t-1}
$$

---

### 3️⃣ Ecuación de la estacionalidad

**Versión aditiva**

$$
s_t = \gamma (Y_t - \ell_t) + (1-\gamma)s_{t-L}
$$

**Versión multiplicativa**

$$
s_t = \gamma \left(\frac{Y_t}{\ell_t}\right) + (1-\gamma)s_{t-L}
$$

---

### 4️⃣ Ecuación de pronóstico

Para $m$ periodos hacia el futuro:

**Versión aditiva**

$$
\hat{Y}_{t+m} = \ell_t + m b_t + s_{t+m-L}
$$

**Versión multiplicativa**

$$
\hat{Y}_{t+m} = (\ell_t + m b_t)\, s_{t+m-L}
$$

---

### 🎯 Interpretación

- El nivel captura la posición actual de la serie.
- La tendencia captura la velocidad de crecimiento o decrecimiento.
- La estacionalidad captura patrones periódicos.
- El modelo combina los tres componentes para generar pronósticos coherentes con la estructura observada en los datos.

## 🔷 ETS: familia de modelos estructurales (Error – Trend – Seasonality)

En la práctica, SES / Holt / Holt‑Winters son **casos particulares** dentro de una familia más general: **ETS**.

### Componentes ETS
- **E (Error):** Aditivo (A) o Multiplicativo (M)  
- **T (Trend):** Ninguna (N), Aditiva (A), o Aditiva amortiguada (Ad)  
- **S (Seasonality):** Ninguna (N), Aditiva (A) o Multiplicativa (M)

Ejemplos útiles:
- **SES** ≈ ETS(A, N, N)
- **Holt (tendencia)** ≈ ETS(A, A, N)
- **Holt‑Winters aditivo** ≈ ETS(A, A, A)
- **Holt‑Winters multiplicativo** ≈ ETS(M, A, M) (si la variabilidad crece con el nivel)

### ¿Aditivo o multiplicativo?
Una regla práctica (que luego validamos con métricas y residuos):
- Si la amplitud de las oscilaciones estacionales es **más o menos constante**, usar **estacionalidad aditiva**.
- Si la amplitud crece cuando la serie crece (efecto “proporcional”), usar **estacionalidad multiplicativa**.

### Tendencia amortiguada (damped)
Para pronósticos a horizontes largos, Holt con tendencia aditiva puede **extrapolar demasiado**.  
La tendencia amortiguada introduce un factor \(\phi\in(0,1)\) que reduce gradualmente el efecto de la tendencia en el futuro.

> En series reales (demanda, producción, matrículas), la tendencia “no suele crecer linealmente para siempre”, por eso ETS con tendencia amortiguada suele ser más realista.

## ✅ Diagnóstico y validación

Un buen pronóstico no es solo "que la curva se vea bonita". Debemos verificar:

1. **Residuos centrados en cero** (sin sesgo sistemático).
2. **No autocorrelación fuerte** en residuos (el modelo capturó la estructura).
3. **Varianza razonable** y ausencia de patrones obvios.
4. **Validación temporal**: separación train/test (o rolling origin) y métricas fuera de muestra.

Estas cuatro verificaciones se implementan en detalle, con código y resultados reales, en la sección **"Diagnóstico de residuos"** del caso aplicado de producción de café, más adelante en este cuadernillo. Allí calculamos explícitamente la media de los residuos, su comportamiento temporal, su histograma, su ACF y la prueba de Ljung–Box.


### 1. MAE (Mean Absolute Error)
$$
MAE = \frac{1}{n} \sum_{t=1}^{n} |Y_t - \hat{Y}_t|
$$
Donde:
-  $Y_t$ es el valor real en el tiempo $ t $.
-  $\hat{Y}_t$ es el valor pronosticado en el tiempo $t$.
- $n$ es el número total de observaciones.

### 2. MSE (Mean Squared Error)
$$
MSE = \frac{1}{n} \sum_{t=1}^{n} (Y_t - \hat{Y}_t)^2
$$
Donde:
- $Y_t$ es el valor real en el tiempo $t$.
- $\hat{Y}_t$ es el valor pronosticado en el tiempo $t$.
- $n$ es el número total de observaciones.

### 3. RMSE (Root Mean Squared Error)
$$
RMSE = \sqrt{\frac{1}{n} \sum_{t=1}^{n} (Y_t - \hat{Y}_t)^2}
$$
Donde:
- $Y_t$ es el valor real en el tiempo $t$.
- $\hat{Y}_t$ es el valor pronosticado en el tiempo $t$.
- $n$ es el número total de observaciones.

### 4. MAPE (Mean Absolute Percentage Error)
$$
MAPE = \frac{100}{n} \sum_{t=1}^{n} \left| \frac{Y_t - \hat{Y}_t}{Y_t} \right|
$$
Donde:
- $Y_t$ es el valor real en el tiempo \( t \).
-  $\hat{Y}_t$ es el valor pronosticado en el tiempo $t$.
- $n$ es el número total de observaciones.

### 5. sMAPE (Symmetric Mean Absolute Percentage Error)
$$
sMAPE = \frac{100}{n}\sum_{t=1}^n \frac{2\,|Y_t-\hat Y_t|}{|Y_t|+|\hat Y_t|}
$$
A diferencia del MAPE, está acotado y es más robusto cuando $Y_t$ toma valores cercanos a cero.

### 6. Sesgo o Error Medio (Bias / ME)
$$
ME = \frac{1}{n}\sum_{t=1}^n (Y_t - \hat Y_t)
$$
- $ME > 0$: el modelo **subestima** sistemáticamente.
- $ME < 0$: el modelo **sobreestima** sistemáticamente.
- $ME \approx 0$: no hay evidencia de sesgo sistemático relevante (pero **no implica** que el modelo sea preciso: errores grandes de signo opuesto pueden cancelarse entre sí).

### 7. MASE (Mean Absolute Scaled Error)
$$
MASE = \frac{MAE_{modelo}}{MAE_{benchmark}}
$$
Para series mensuales con estacionalidad, el benchmark natural es el **Seasonal Naïve**. Interpretación:

- $MASE < 1$: el modelo supera al benchmark simple.
- $MASE = 1$: desempeño equivalente al benchmark.
- $MASE > 1$: el benchmark simple pronostica mejor que el modelo — una señal de alerta importante.

## 🎯 Benchmark: ¿el modelo realmente pronostica mejor?

Antes de examinar Holt, Holt amortiguado o Holt‑Winters, necesitamos un punto de comparación honesto. Un modelo complejo que no supera a un pronóstico ingenuo **no aporta valor práctico**, sin importar qué tan sofisticado sea.

### Naïve
$$
\hat{Y}_{t+h\mid t} = Y_t
$$
El pronóstico es simplemente el último valor observado, repetido para todo el horizonte.

### Seasonal Naïve
$$
\hat{Y}_{t\mid t-12} = Y_{t-12}
$$
Cada mes se pronostica usando el valor observado en el **mismo mes del año anterior**. Para una serie **mensual con estacionalidad marcada**, como la producción de café, este es el benchmark relevante — un modelo que no logre superarlo no está aportando nada más allá de repetir el patrón estacional ya conocido.

También incluimos el **promedio histórico** como referencia adicional (menos exigente, útil solo como piso de comparación).


````python
def future_index(train: pd.Series, h: int) -> pd.DatetimeIndex:
    '''Índice mensual (MS) para los h periodos siguientes al final de train.'''
    return pd.date_range(start=train.index[-1] + pd.DateOffset(months=1), periods=h, freq="MS")


def naive_forecast(train: pd.Series, h: int) -> pd.Series:
    '''Y_hat(t+h) = Y_t (ultimo valor observado, repetido).'''
    return pd.Series(train.iloc[-1], index=future_index(train, h))


def seasonal_naive_forecast(train: pd.Series, h: int, m: int = 12) -> pd.Series:
    '''Y_hat_t = Y_(t-m) (repite el ultimo ciclo estacional completo).'''
    last_season = train.iloc[-m:]
    reps = int(np.ceil(h / m))
    vals = np.tile(last_season.values, reps)[:h]
    return pd.Series(vals, index=future_index(train, h))


def mean_forecast(train: pd.Series, h: int) -> pd.Series:
    '''Promedio histórico de train, repetido para todo el horizonte.'''
    return pd.Series(train.mean(), index=future_index(train, h))


fc_naive = naive_forecast(train, h).reindex(test.index)
fc_snaive = seasonal_naive_forecast(train, h, m=12).reindex(test.index)
fc_mean = mean_forecast(train, h).reindex(test.index)

plt.figure(figsize=(11, 4))
plt.plot(train.index[-24:], train.values[-24:], label="Train (últimos 24m)")
plt.plot(test.index, test.values, label="Test (real)", color="black", linewidth=2)
plt.plot(fc_naive.index, fc_naive.values, "--", label="Naïve")
plt.plot(fc_snaive.index, fc_snaive.values, "--", label="Seasonal Naïve")
plt.plot(fc_mean.index, fc_mean.values, "--", label="Promedio histórico")
plt.title("Benchmarks de pronóstico sobre el conjunto de test")
plt.xlabel("Fecha")
plt.ylabel("Miles de sacos de 60 kg")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

### Métricas de evaluación (ampliadas)

Mantenemos MAE, RMSE, MAPE y sMAPE, y agregamos **Bias (ME)** y **MASE** (ver definiciones más arriba). El MASE se calcula dividiendo el MAE del modelo entre el MAE del **Seasonal Naïve** medido sobre el mismo conjunto de evaluación.

También generalizamos el ajuste de modelos en una sola función `fit_model`, que:
- admite benchmarks y modelos ETS (SES, Holt, Holt amortiguado, HW aditivo/multiplicativo) con la misma interfaz;
- **captura explícitamente** los `ConvergenceWarning` en lugar de ocultarlos;
- **no fuerza** el ajuste multiplicativo si la serie no es estrictamente positiva — explica el motivo en su lugar.


````python
import warnings
from statsmodels.tools.sm_exceptions import ConvergenceWarning
from statsmodels.tsa.holtwinters import SimpleExpSmoothing, ExponentialSmoothing


def forecast_metrics(y_true: pd.Series, y_pred: pd.Series, mae_benchmark: float = None) -> pd.Series:
    '''Métricas de error para pronóstico univariado.

    mae_benchmark: MAE del modelo de referencia (Seasonal Naïve) sobre el
    mismo conjunto de evaluación, usado para calcular el MASE. Si es None,
    MASE se reporta como NaN.
    '''
    y_true = y_true.astype(float)
    y_pred = y_pred.astype(float)
    err = y_true - y_pred

    me = np.mean(err)
    mae = np.mean(np.abs(err))
    rmse = np.sqrt(np.mean(err ** 2))

    denom = np.where(np.abs(y_true) < 1e-12, np.nan, np.abs(y_true))
    mape = np.nanmean(np.abs(err / denom)) * 100

    smape_denom = np.abs(y_true) + np.abs(y_pred)
    smape_denom = np.where(smape_denom < 1e-12, np.nan, smape_denom)
    smape = np.nanmean(2.0 * np.abs(err) / smape_denom) * 100

    mase = mae / mae_benchmark if (mae_benchmark is not None and mae_benchmark > 0) else np.nan

    return pd.Series({
        "Bias_ME": me, "MAE": mae, "RMSE": rmse,
        "MAPE_%": mape, "sMAPE_%": smape, "MASE": mase,
    })


def fit_model(train: pd.Series, h: int, kind: str, seasonal_periods: int = 12):
    '''Ajusta un modelo (o calcula un benchmark) y devuelve (forecast, aic, convergio, mensaje).

    kind in {"naive", "seasonal_naive", "mean", "ses", "holt", "holt_damped", "hw_add", "hw_mul"}
    '''
    aic, converged, msg = np.nan, True, ""

    if kind == "naive":
        return naive_forecast(train, h), aic, converged, msg
    if kind == "seasonal_naive":
        return seasonal_naive_forecast(train, h, seasonal_periods), aic, converged, msg
    if kind == "mean":
        return mean_forecast(train, h), aic, converged, msg

    try:
        with warnings.catch_warnings(record=True) as wlist:
            warnings.simplefilter("always", ConvergenceWarning)

            if kind == "ses":
                mdl = SimpleExpSmoothing(train, initialization_method="estimated").fit(optimized=True)
            elif kind == "holt":
                mdl = ExponentialSmoothing(train, trend="add", seasonal=None,
                                            initialization_method="estimated").fit(optimized=True)
            elif kind == "holt_damped":
                mdl = ExponentialSmoothing(train, trend="add", damped_trend=True, seasonal=None,
                                            initialization_method="estimated").fit(optimized=True)
            elif kind == "hw_add":
                mdl = ExponentialSmoothing(train, trend="add", seasonal="add", seasonal_periods=seasonal_periods,
                                            initialization_method="estimated").fit(optimized=True)
            elif kind == "hw_mul":
                if (train <= 0).any():
                    return None, np.nan, False, "No aplicable: la estacionalidad multiplicativa requiere valores estrictamente positivos."
                mdl = ExponentialSmoothing(train, trend="mul", seasonal="mul", seasonal_periods=seasonal_periods,
                                            initialization_method="estimated").fit(optimized=True)
            else:
                raise ValueError(f"kind desconocido: {kind}")

            converged = not any(issubclass(w.category, ConvergenceWarning) for w in wlist)

        fc = mdl.forecast(h)
        aic = getattr(mdl, "aic", np.nan)
        return fc, aic, converged, msg

    except Exception as e:
        return None, np.nan, False, f"Error de ajuste: {e}"
````

### Comparación en el conjunto de test (todos los candidatos)

Ajustamos **todos** los modelos candidatos sobre `train` y los evaluamos sobre el mismo `test`, junto con los benchmarks ya calculados: Seasonal Naïve, Naïve, SES, Holt, Holt amortiguado, Holt‑Winters aditivo y Holt‑Winters multiplicativo.

Ordenamos por MASE y luego por RMSE, pero la tabla se usa para **describir**, no para elegir automáticamente un ganador.


````python
candidate_kinds = {
    "Naive": "naive",
    "Seasonal Naive": "seasonal_naive",
    "SES": "ses",
    "Holt": "holt",
    "Holt amortiguado": "holt_damped",
    "HW aditivo": "hw_add",
    "HW multiplicativo": "hw_mul",
}

rows = []
forecasts = {}

for name, kind in candidate_kinds.items():
    fc, aic, converged, msg = fit_model(train, h, kind)
    if fc is None:
        print(f"{name}: {msg}")
        continue
    fc = fc.reindex(test.index)
    forecasts[name] = fc
    rows.append({"Modelo": name, "AIC": aic, "Convergió": converged})

tabla_simple = pd.DataFrame(rows).set_index("Modelo")

# El MASE se calcula con el MAE del Seasonal Naïve como denominador
mae_bench = forecast_metrics(test, forecasts["Seasonal Naive"])["MAE"]

metricas = {name: forecast_metrics(test, fc, mae_benchmark=mae_bench) for name, fc in forecasts.items()}
tabla_metricas = pd.DataFrame(metricas).T
tabla_simple = tabla_simple.join(tabla_metricas)

tabla_simple.sort_values(["MASE", "RMSE"])
````

### 2) Re‑entrenar para pronóstico final (dejando **solo `train`**)

Una práctica común es:

1) Validar con train/test y backtesting (ver secciones siguientes).
2) **Seleccionar deliberadamente** un modelo para producción, sopesando MASE, RMSE, Bias, diagnóstico de residuos, estabilidad y convergencia — **no** solo la métrica más baja de una única partición.
3) **Re‑entrenar con toda la serie disponible** para producir el pronóstico final.

Para este cuadernillo usamos **Holt‑Winters aditivo** como modelo de producción, como **ejemplo de decisión razonada**: en el backtesting (más abajo) su desempeño es prácticamente equivalente al de Holt‑Winters multiplicativo, pero este último muestra más problemas de convergencia entre orígenes — una razón operacional válida para preferir el modelo aditivo aunque su RMSE puntual no sea el más bajo.

Después de re‑entrenar, dejamos `train` como la **serie completa** (y removemos `test` para evitar confusiones).


````python
full = y.copy()

with warnings.catch_warnings(record=True) as wlist:
    warnings.simplefilter("always", ConvergenceWarning)
    final_model = ExponentialSmoothing(
        full, trend="add", seasonal="add", seasonal_periods=12,
        initialization_method="estimated"
    ).fit(optimized=True)
    final_converged = not any(issubclass(w.category, ConvergenceWarning) for w in wlist)

final_fc_12 = final_model.forecast(12)
print(f"AIC: {final_model.aic:.2f} | Convergió: {final_converged}")

# Dejar SOLO train (train = serie completa)
train = full
try:
    del test
except NameError:
    pass

plt.figure(figsize=(11, 4))
plt.plot(train.index, train.values, label="Serie (train completo)")
plt.plot(final_fc_12.index, final_fc_12.values, label="Pronóstico 12 meses")
plt.title("Pronóstico final (modelo re-entrenado con toda la serie)")
plt.xlabel("Fecha")
plt.ylabel("Miles de sacos de 60 kg")
plt.legend()
plt.show()

train.tail(), final_fc_12.head()
````

### 3) Backtesting con **expanding window** (ventana creciente) — todos los candidatos

En la sección anterior evaluamos una sola partición train/test. Ahora aplicamos backtesting de ventana creciente a **todos los modelos candidatos** (benchmarks incluidos), no solo al modelo elegido — así podemos verificar si las conclusiones anteriores son estables en el tiempo.

- Entrenamos con una ventana inicial.
- Pronosticamos un horizonte fijo (**12 meses**).
- Avanzamos el origen y repetimos (la ventana **crece**, no se desplaza — no usamos particiones aleatorias).
- Registramos, además del error agregado, el error **por horizonte** ($h=1,\dots,12$) y si cada ajuste **convergió**.


````python
from typing import Dict


def expanding_window_backtest_multi(
    y: pd.Series,
    kinds: Dict[str, str],
    initial_train_end: str,
    horizon: int = 12,
    step: int = 12,
    seasonal_periods: int = 12,
):
    '''Backtesting de ventana creciente para varios modelos/benchmarks a la vez.

    Devuelve:
    - conv_df: registro de convergencia por (modelo, origen)
    - horizon_df: error por (modelo, origen, horizonte h)
    - global_df: métricas agregadas por modelo (concatenando todos los orígenes)
    '''
    y = y.asfreq("MS").dropna()
    all_dates = y.index

    start_idx = np.where(all_dates == pd.Timestamp(initial_train_end))[0]
    if len(start_idx) == 0:
        raise ValueError("initial_train_end no coincide con el índice de la serie.")
    start_idx = int(start_idx[0])
    last_origin_idx = len(all_dates) - horizon - 1

    rows_conv, rows_h = [], []
    preds_all = {name: [] for name in kinds}
    trues_all = {name: [] for name in kinds}

    for origin_idx in range(start_idx, last_origin_idx + 1, step):
        train_end_date = all_dates[origin_idx]
        train_y = y.loc[:train_end_date]
        test_y = y.iloc[origin_idx + 1: origin_idx + 1 + horizon]

        for name, kind in kinds.items():
            fc, aic, converged, msg = fit_model(train_y, horizon, kind, seasonal_periods)
            rows_conv.append({"Modelo": name, "origen": train_end_date,
                               "convergio": converged, "aic": aic, "mensaje": msg})
            if fc is None:
                continue
            fc = fc.reindex(test_y.index)
            preds_all[name].append(fc.rename(train_end_date))
            trues_all[name].append(test_y.rename(train_end_date))
            for h_i in range(1, horizon + 1):
                if h_i - 1 < len(test_y):
                    y_t, y_p = test_y.iloc[h_i - 1], fc.iloc[h_i - 1]
                    rows_h.append({"Modelo": name, "origen": train_end_date, "h": h_i,
                                    "abs_err": abs(y_t - y_p), "sq_err": (y_t - y_p) ** 2})

    conv_df = pd.DataFrame(rows_conv)
    horizon_df = pd.DataFrame(rows_h)

    global_rows = []
    for name in kinds:
        if len(preds_all[name]) == 0:
            continue
        y_pred_all = pd.concat(preds_all[name])
        y_true_all = pd.concat(trues_all[name])
        global_rows.append({"Modelo": name, **forecast_metrics(y_true_all, y_pred_all).to_dict()})
    global_df = pd.DataFrame(global_rows).set_index("Modelo")

    return conv_df, horizon_df, global_df


# Mismos modelos/benchmarks candidatos de la sección anterior
kinds_backtest = candidate_kinds
kinds_backtest
````

````python
# Ventana inicial: ~5 años (60 meses), para observar varias temporadas anuales
initial_train_end = y.index[60]

conv_df, horizon_df, global_df = expanding_window_backtest_multi(
    y=y,
    kinds=kinds_backtest,
    initial_train_end=str(initial_train_end.date()),
    horizon=12,
    step=12,
)

mae_bench_bt = global_df.loc["Seasonal Naive", "MAE"]
global_df["MASE"] = global_df["MAE"] / mae_bench_bt

global_df.sort_values(["MASE", "RMSE"])
````

**Tabla final de backtesting.** Ordenamos por MASE y luego por RMSE solo para facilitar la lectura — **no** para declarar automáticamente un ganador. La selección final debe considerar simultáneamente:

- **MASE**: ¿el modelo supera al Seasonal Naïve de forma consistente en múltiples orígenes?
- **RMSE**: magnitud del error, sensible a errores grandes.
- **Bias**: ¿hay sub- o sobreestimación sistemática?
- **Diagnóstico de residuos** (sección anterior).
- **Estabilidad y convergencia** (siguiente tabla).

### Control de convergencia

No ocultamos las advertencias de convergencia: las registramos explícitamente durante el backtesting.


````python
resumen_convergencia = conv_df.groupby("Modelo").agg(
    Ajustes_realizados=("convergio", "size"),
    Problemas_de_convergencia=("convergio", lambda s: int((~s).sum())),
)
resumen_convergencia
````

Un modelo que ocasionalmente exhibe mejor RMSE pero presenta **problemas frecuentes de convergencia** puede ser menos confiable operacionalmente: en producción, un ajuste que no converge de forma consistente es un riesgo, no un detalle técnico menor.


## 📏 Error según el horizonte de pronóstico

No basta con una métrica agregada a 12 meses: un modelo puede ser excelente a 1–3 meses y deteriorarse notablemente a 9–12 meses.


````python
tabla_horizonte = (
    horizon_df.groupby(["Modelo", "h"])
    .agg(MAE=("abs_err", "mean"), RMSE=("sq_err", lambda s: np.sqrt(s.mean())))
    .reset_index()
)

tabla_horizonte.pivot(index="h", columns="Modelo", values="RMSE").round(1)
````

````python
modelos_plot = ["Seasonal Naive", "Holt", "HW aditivo", "HW multiplicativo"]

plt.figure(figsize=(11, 5))
for m in modelos_plot:
    sub = tabla_horizonte[tabla_horizonte["Modelo"] == m]
    plt.plot(sub["h"], sub["RMSE"], marker="o", label=m)

plt.title("RMSE según horizonte de pronóstico (backtesting, ventana creciente)")
plt.xlabel("Horizonte h (meses)")
plt.ylabel("RMSE")
plt.xticks(range(1, 13))
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

Un modelo puede funcionar muy bien a uno o tres meses y deteriorarse considerablemente cuando intenta pronosticar nueve o doce meses — esto es información tan importante como el error agregado.

## 🔷 De Holt‑Winters a la familia ETS

Hasta ahora trabajamos con `statsmodels.tsa.holtwinters.ExponentialSmoothing`. Es común llamar informalmente "ETS" a cualquier modelo de suavizamiento exponencial, pero conviene ser precisos: **ETS(E,T,S)** es una familia más general definida por tres letras:

- **E (Error):** aditivo o multiplicativo.
- **T (Trend):** ninguna, aditiva o aditiva amortiguada.
- **S (Seasonality):** ninguna, aditiva o multiplicativa.

`statsmodels` también ofrece una implementación explícita de esta familia: `statsmodels.tsa.exponential_smoothing.ets.ETSModel`. Su principal ventaja pedagógica aquí es que provee **intervalos de pronóstico** de forma nativa, algo que el método clásico de Holt‑Winters no entrega directamente por punto.

### Pronóstico puntual vs. incertidumbre

Un pronóstico puntual sin incertidumbre es información incompleta. A continuación mostramos el pronóstico del modelo final junto con bandas de predicción del 80% y del 95%.


````python
from statsmodels.tsa.exponential_smoothing.ets import ETSModel

ets_model = ETSModel(
    train, error="add", trend="add", seasonal="add", seasonal_periods=12,
    initialization_method="estimated"
).fit(disp=False)

pred = ets_model.get_prediction(start=len(train), end=len(train) + 11)
bandas_80 = pred.summary_frame(alpha=0.20)
bandas_95 = pred.summary_frame(alpha=0.05)

plt.figure(figsize=(11, 5))
plt.plot(train.index[-36:], train.values[-36:], label="Serie histórica (últimos 36m)", color="black")
plt.plot(bandas_95.index, bandas_95["mean"], label="Pronóstico puntual", color="tab:blue")

plt.fill_between(bandas_95.index, bandas_95["pi_lower"], bandas_95["pi_upper"],
                  alpha=0.15, color="tab:blue", label="Intervalo 95%")
plt.fill_between(bandas_80.index, bandas_80["pi_lower"], bandas_80["pi_upper"],
                  alpha=0.30, color="tab:blue", label="Intervalo 80%")

plt.title("Pronóstico ETS(A,A,A) con intervalos de predicción")
plt.xlabel("Fecha")
plt.ylabel("Miles de sacos de 60 kg")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

El intervalo se **amplía a medida que aumenta el horizonte**, reflejando que la incertidumbre acumulada crece cuanto más lejos pronosticamos. Esto es consistente con lo observado en la sección de error por horizonte: a mayor $h$, mayor error esperado y mayor incertidumbre.

> No usamos $\pm 1.96 \times RMSE$ como atajo para construir estos intervalos: esa fórmula asume errores independientes e idénticamente distribuidos con varianza constante, un supuesto que no se sostiene automáticamente en pronósticos multi‑paso. Los intervalos anteriores provienen del modelo de espacio de estados ETS, que propaga correctamente la incertidumbre según la estructura del modelo.

## 🧭 Flujo recomendado de modelamiento

```text
1. Visualizar la serie
        ↓
2. Identificar nivel, tendencia y estacionalidad
        ↓
3. Construir baseline (Naïve / Seasonal Naïve)
        ↓
4. Ajustar modelos candidatos (SES, Holt, Holt amortiguado, HW aditivo/multiplicativo)
        ↓
5. Revisar residuos (media, ACF, Ljung-Box)
        ↓
6. Realizar train/test temporal
        ↓
7. Realizar rolling-origin / expanding-window
        ↓
8. Comparar MAE, RMSE, Bias y MASE
        ↓
9. Analizar error según horizonte
        ↓
10. Revisar estabilidad y convergencia
        ↓
11. Seleccionar modelo (criterio múltiple, no una sola métrica)
        ↓
12. Reentrenar con toda la información
        ↓
13. Generar pronóstico final (con intervalos de incertidumbre)
```

> **Conclusión.** El mejor modelo de pronóstico no es necesariamente el que mejor ajusta el pasado; es aquel que demuestra capacidad predictiva, ausencia de sesgos importantes, residuos razonables y desempeño estable frente a alternativas simples, a través del tiempo y de diferentes horizontes de pronóstico.
